Tipos de variables en python

1. Variables tipo string

```python
nombre = "Jose Luis"
print (nombre)
print (type(nombre))
```

2. Variable tipo numero entero

```python
edad = 18
print (edad)
print(type(edad))
```

3. Variable tipo numero decimal

```python
estatura = 1.8
print(estatura)
print(type(estatura))
```

4. Variable tipo booleano

```python
es_mayor_de_edad = True
es_menor_de_edad = False
print(es_mayor_de_edad)
print(type(es_mayor_de_edad))
```
